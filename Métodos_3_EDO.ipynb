```python
import numpy as np

def Euler(f, t0, tf, y0, h):
  t = np.arange(t0, tf+h, h)
  y = np.zeros(len(t))

  for i in range(len(t) - 1):
    y[i+1] = y + h * f([t(i)], [y(i)])

    return t, y
```
